```python
a =  input()
check = True
j = len(a) -1
i = 0
while i < j:
 if a[i] != a[j]:
   check = False
   break
 i = i + 1
 j = j - 1
if check:
  print("Palindrome")
else:
  print("Not Palindrome")
```

```python
lst = ['Kunal', 45, 89, ['Rajeev','Asto']]
print(lst)

#append(value)
lst.append(['Amber', 3,4])
print(lst)

#extend(iterable)
lst.extend([10,11])
print(lst)

#insert(index,value)
lst.insert(3,'ayush')
print(lst)

#pop()
lst.pop()
print(lst)

#remove(value)
lst.remove(89)
print(lst)

```

```python
weekly_sales = {}

while True:
    item = input().strip().lower()

    if item == "end":
        break

    quantity = int(input(f"How many {item}s were sold?: "))

    if item in weekly_sales:
        weekly_sales[item] += quantity
    else:
        weekly_sales[item] = quantity
    print(f"Recorded: +{quantity} {item}(s)\n")


if not weekly_sales:
    print("No items were sold this week.")
else:
    for item, total_sold in weekly_sales.items():
        print(f"• {item.capitalize()}: {total_sold} sold")
```

```python
stock = np.random.randint(20, 80,(5,5))
print(stock)

print("\n")

demand = np.random.randint(20, 80, (5, 5))
print(demand)

print("\n")

#remaining stock:

remaining_stock = stock -  demand
print(remaining_stock)
print("\n")

#neg_values -> 0

replace = np.where(remaining_stock < 0, 0, remaining_stock)
print(replace)
print("\n")

#total shortage:
total_shortage = np.sum(np.where(remaining_stock < 0, -remaining_stock, 0))
print(total_shortage)
print("\n")

#products with zero stock:
zero_stock = np.sum(remaining_stock < 0)
print(zero_stock)
print("\n")

#sort each row based on remaing stock
sort_row = np.sort(remaining_stock, axis = 1)
print(sort_row)
print("\n")
```

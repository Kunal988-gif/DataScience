```python
a = np.random.randint(2, 15, 6)
print(a)
print(np.min(a))
print(np.max(a))
print(np.mean(a))
print(np.median(a))
print(np.std(a))

b = a.reshape((3, 2))
print(b)

```

```python
a = np.array([2, 3, 5 , 7])
b = np.array([3 , 6, 8 ,2])

concat = np.concatenate((a, b))
print(concat)
concat = np.transpose((a, b))
print(concat)
```

```python
a = np.random.randint(2, 15, 6)
b = np.random.randint(1, 19, 6)

print(a)
print(b)

c = np.where(b > 2, b , a)
print(c)
```

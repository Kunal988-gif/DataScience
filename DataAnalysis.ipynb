**GOOD LOAN METRICS ANALYSIS**

```python
number_of_good_loan = df[df['loan_status'] == 'Fully Paid']['id'].nunique()
print(number_of_good_loan)
```

```python
good_loan = df.groupby('loan_status').get_group('Fully Paid')
good_loan_amount = good_loan['loan_amount'].sum()
print(good_loan_amount)
```

```python
loan_status_data = df.groupby('loan_status').agg({
    'id': 'count',
    'loan_amount': 'sum',
    'total_payment': 'sum',
    'dti': 'mean' ,
    'int_rate': 'mean'

})

loan_status_data.rename(columns={'id' : 'Number of Applications'},inplace = True)
display(loan_status_data)
```

# **GRAPH ANALYSIS**

**Relation Between Issue Month vs Loan Amount**

```python
df['month'] = df['issue_date'].dt.month

new_df = df.groupby('month',as_index = False)['loan_amount'].sum()
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']

plt.plot(months,new_df['loan_amount'],marker = 'o')
plt.grid()
plt.title('Relation Between Issue Month vs Loan Amount')
plt.show()
```

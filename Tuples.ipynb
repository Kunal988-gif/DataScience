```python
tp = (1, 2, 9 , 0, 5, 5, 5,7)

newTup = sorted(tp)
print(newTup) #returns a new List

newTup = tuple(sorted(tp))
print(newTup) #converts return list from sorted to a tuple


#tuples Funtions

#index()
targetIndex = tp.index(9)
print(targetIndex)

#count(element)
tp.count(5)

#sum()
total = sum(tp)
print(total)

#max()
maxNum = max(tp)
print(maxNum)

#min()
minNum = min(tp)
print(minNum)
```
